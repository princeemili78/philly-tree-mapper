```python
import geopandas as gpd
from pathlib import Path
import matplotlib as plt

RAW_DATA_DIR = Path("../data/raw")
PROCESSED_DATA_DIR = Path("../data/processed")

PARKS_FILE = RAW_DATA_DIR / "PPR_Properties.geojson"
TREES_FILE = RAW_DATA_DIR / "ppr_tree_inventory_2025.geojson"
OUTPUT_FILE = PROCESSED_DATA_DIR / "trees_in_parks.geojson"
```

```python
parks = gpd.read_file(PARKS_FILE)
trees = gpd.read_file(TREES_FILE)
```

```python
num_trees_by_type = trees["tree_name"].value_counts().reset_index().set_index("tree_name")
num_trees_by_type
```

```python
ax = num_trees_by_type.plot(kind='bar', rot=0)
ax.set_title('Number per type of tree')
ax.set_xlabel('Tree name')
ax.set_ylabel('Count')
ax.plot
```

```python
# Get top 10 trees and make bar chart
top_10_trees = num_trees_by_type.head(10)
ax = top_10_trees.plot(kind='barh')
ax.set_title('Number per type of tree')
ax.set_xlabel('Tree name')
ax.set_ylabel('Count')
ax.plot
```

```python
# Get top 20 trees and make bar chart
top_20_trees = num_trees_by_type.head(20)
ax = top_20_trees.plot(kind='barh')
ax.set_title('Number per type of tree')
ax.set_xlabel('Tree name')
ax.set_ylabel('Count')
ax.plot
```

```python
# Get top 30 trees and make bar chart
top_30_trees = num_trees_by_type.head(30)
ax = top_30_trees.plot(kind='barh')
ax.set_title('Number per type of tree')
ax.set_xlabel('Tree name')
ax.set_ylabel('Count')
ax.plot
```

```python
# How many trees have above 10 trees in the city
trees_above_5 = num_trees_by_type[num_trees_by_type["count"] <= 500]
print(len(trees_above_5))
```
