```python
import geopandas as gpd
from pathlib import Path

RAW_DATA_DIR = Path("../data/raw")
PROCESSED_DATA_DIR = Path("../data/processed")

PARKS_FILE = RAW_DATA_DIR / "PPR_Properties.geojson"
TREES_FILE = RAW_DATA_DIR / "ppr_tree_inventory_2025.geojson"
OUTPUT_FILE = PROCESSED_DATA_DIR / "trees_in_parks.geojson"

```

```python
parks = gpd.read_file(PARKS_FILE)
trees = gpd.read_file(TREES_FILE)
```

```python
parks_cleaned = parks.dropna()
```

```python
parks_cleaned.isna().sum()
```

```python
trees_in_parks = gpd.sjoin(trees, parks_cleaned, how='inner', predicate='within')
```

```python
trees_in_parks.isna().sum()
```

```python
named_trees_in_parks = trees_in_parks.dropna(subset=["tree_name"])
```

```python
num_trees_per_park = named_trees_in_parks.groupby(["official_name", "tree_name"]).size().reset_index()
num_trees_per_park
```
