```python
import pandas as pd
import matplotlib as plt
from pathlib import Path
PROCESSED_DATA_DIR = Path("../data/processed")
TREE_ACTIVITY_FILE = PROCESSED_DATA_DIR / "species_activity_curves.csv"
GENUS_ACTIVITY_FILE = PROCESSED_DATA_DIR / "genus_activity_curves.csv"
```

```python
tree_activity = pd.read_csv(TREE_ACTIVITY_FILE)
genus_activity = pd.read_csv(GENUS_ACTIVITY_FILE)
```

```python
leaves_colored_acer_rubrum = tree_activity[(tree_activity["scientific_name"] == "Acer rubrum") & (tree_activity["Phenophase_ID"] == 498)]
leaf_color_plot = leaves_colored_acer_rubrum.drop(columns=["scientific_name", "Phenophase_ID"])
leaf_color_plot.T.plot(
    kind="line",
    xlabel="Day of Year",
    ylabel="Percent Observed",
    title="When do the leaves change color?",
    legend=False
)
```

```python
flowering_acer_rubrum = tree_activity[(tree_activity["scientific_name"] == "Acer rubrum") & (tree_activity["Phenophase_ID"] == 186)]
acer_flowering_plot = flowering_acer_rubrum.drop(columns=["scientific_name", "Phenophase_ID"])
acer_flowering_plot.T.plot(
    kind="line",
    xlabel="Day of Year",
    ylabel="Percent Observed",
    title="When do the flowers come out",
    legend=False
)
```

```python
leaves_falling_acer_rubrum = tree_activity[(tree_activity["scientific_name"] == "Acer rubrum") & (tree_activity["Phenophase_ID"] == 471)]
acer_leaves_falling_plot = leaves_falling_acer_rubrum.drop(columns=["scientific_name", "Phenophase_ID"])
acer_leaves_falling_plot.T.plot(
    kind="line",
    xlabel="Day of Year",
    ylabel="Percent Observed",
    title="When do the leaves fall",
    legend=False
)
```
