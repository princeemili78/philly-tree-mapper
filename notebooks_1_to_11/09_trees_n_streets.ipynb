```python
import geopandas as gpd
import pandas as pd
import folium
import xyzservices
from IPython.display import display, HTML
pd.set_option("display.max_columns", None)
```

```python
trees = gpd.read_file("../data/processed/philly_trees.geojson").to_crs(epsg=2272)
streets = gpd.read_file("../data/processed/streets.geojson").to_crs(epsg=2272)
parks = gpd.read_file("../data/processed/parks.geojson")
```

```python
st_trees = gpd.sjoin_nearest(
    trees,
    streets,
    max_distance=100,
    distance_col="dist_street",
    how="left"
)
print(f'tree shape {trees.shape}. street tree shape {st_trees.shape}')

```

```python
free_trees = st_trees.loc[st_trees["index_right"].isna()]
print(f"There are {len(free_trees)} free trees")
```

```python
street_trees = st_trees[~st_trees["index_right"].isna()]
```

```python
def plot_st_vs_fr(species, s_gdf, f_gdf):
    street = s_gdf.loc[s_gdf["scientific_name"] == species]
    free = f_gdf.loc[f_gdf["scientific_name"] == species]
    m = free.to_crs(epsg=4326).explore(
        tiles=xyzservices.providers.Esri.NatGeoWorldMap,
        location=[39.99, -75.13],
        zoom_start=11,
        marker_kwds={"radius": .2, "fill": True, "color": "green"},
        name="Free trees"
    )
    street.to_crs(epsg=4326).explore(
        m=m,
        marker_kwds={"radius": .2, "fill": True, "color": "black"},
        name="street trees"
    )
    folium.LayerControl().add_to(m)
    return m
```

```python
st_prk_trs = gpd.sjoin(
street_trees.drop(columns=["index_right"]),
parks,
predicate="within",
how="inner"
)


f_prk_trs = gpd.sjoin(
    free_trees.drop(columns=["index_right"]),
    parks,
    predicate="within",
    how="inner"
)
print(f"there are {len(st_prk_trs)} street trees in parks, and {len(f_prk_trs)} free trees in parks")
print(f"{len(f_prk_trs)/ len(free_trees)}% free trees in parks, {len(st_prk_trs)/len(street_trees)}% street trees in parks")


parkless_street_trees = street_trees[~street_trees["objectid"].isin(st_prk_trs["objectid"])]

free_park_trees = pd.concat([free_trees, st_prk_trs])
print(f"{parkless_street_trees["objectid"].nunique() + free_park_trees["objectid"].nunique()} = {len(trees)}")
```
