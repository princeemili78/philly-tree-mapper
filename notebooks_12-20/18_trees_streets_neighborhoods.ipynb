```python
import geopandas as gpd
```

```python
trees = gpd.read_file("/Users/prince/philly-tree-mapper/data/processed/philly_trees.geojson").to_crs(epsg=2272)
streets = gpd.read_file("/Users/prince/philly-tree-mapper/data/processed/streets.geojson").to_crs(epsg=2272)
parks = gpd.read_file("/Users/prince/philly-tree-mapper/data/processed/parks.geojson").to_crs(epsg=2272)
hoods = gpd.read_file("/Users/prince/philly-tree-mapper/data/raw/MUSA_zillow_neighborhoods.geojson").to_crs(epsg=2272)
```

```python
# Add philly location data to tree data
trees = trees.drop_duplicates(subset="objectid")
gdfs = [trees, streets, parks, hoods]
tree_spots = gdfs[0]

for gdf in gdfs[1:]:
    if "stname" in list(gdf.columns):
        tree_spots = gpd.sjoin_nearest(
        tree_spots,
        gdf,
        max_distance=100,
        distance_col="dist_street",
        how="left"
    )
    else:
        tree_spots = gpd.sjoin(
        tree_spots.drop(columns=["index_right"]),
        gdf,
        predicate="within",
        how="left"
        )
```

```python
tree_spots = tree_spots.drop(columns="index_right")
```
