```python
def mobot_flower_visible(desc):
    if desc == "N/a":
           return np.nan
    elif desc == "Insignificant" or desc == "Fragrant, Insignificant":
        return False
    else:
        return True
```

```python
def usda_flower_visible(desc):
    if desc != desc:
        return np.nan
    elif desc == "No":
        return False
    else:
        return True
```

```python
def flower_visible(row):
    mobot = mobot_flower_visible(row["Flower"])
    usda = usda_flower_visible(row["Morphology/Physiology | Flower Conspicuous"])
    if mobot is True or usda is True:
        return True
    elif mobot != mobot and usda != usda:
        return np.nan
    else:
        return False

        
```

```python
phenos["Visible_Flowers"] = phenos.apply(lambda row: flower_visible(row), axis=1)
```

STARTING POINT FOR TOMORROW. ADD BACK TYPE TO RESCRAPE MOBOT FOR TYPE OF TREE IT IS.

```python

# For moderates, leaf retention is a good indicator of whether or not it will be bare.
# Moderate porosity + Leaf Retention == "No" -> Decidious tree, will be bare
# Moderate porosity + Leaf Retention == "Yes" -> Evergreen tree, all leaves there
    #Notable exception is Magnolia virginiana, semi evergreen. Loses leaves depending on climate, will treat it as decidous in philly
    #This is kinda hacky
def usda_winter_leaves(row):
    """Put in row of pheno df and return True if leaves in winter, and false if no leaves in winter
    return NaN if inconclusive or no data"""
    leaf_retention = row["Morphology/Physiology | Leaf Retention"]
    winter_foliage = row["Morphology/Physiology | Foliage Porosity Winter"]
    name = row["scientific_name"]
    if name == "magnolia virginiana":
        return False
    elif leaf_retention != leaf_retention or winter_foliage != winter_foliage:
        return np.nan
    elif winter_foliage == "Dense":
        return True
    elif winter_foliage == "Porous":
        return False
    elif winter_foliage == "Moderate" and leaf_retention == "No":
        return False
    elif winter_foliage == "Moderate" and leaf_retention == "Yes":
        return True
    
```

```python
def mobot_winter_leaves(row):
    """Put in row of pheno df and return true if there are leaves in the winter
    false if not, return NaN if inconclusive"""
    tree_type = row["Type"]
    leaf = row["Leaf"]
    mobot_match = row["MOBOT match"]
    if mobot_match is False:
        return np.nan
    elif tree_type == "Broadleaf evergreen" or tree_type == "Needled evergreen":
        return True
    elif leaf == "Colorful, Evergreen"  or leaf == "Evergreen" or leaf == "Fragrant, Evergreen":
        return True
    else:
        return np.nan
```

```python
phenos.loc[phenos["MOBOT match"] == True]
```

```python
def winter_leaves(row):
    """Takes one row of pheno df and returns True if it has leaves in winter, false if not
    NaN if inconclusive. Does this by using MOBOT and USDA data"""
    mobot = mobot_winter_leaves(row)
    usda = usda_winter_leaves(row)
    if mobot is True or usda is True:
        return True
    elif mobot != mobot and usda != usda:
        return np.nan
    else:
        return False
    
```

```python
phenos["Winter_Leaves"] = phenos.apply(lambda row: winter_leaves(row), axis=1)
```

```python
# Find if shit change colors in the fall
def usda_fall_conspic(row):
    """Function takes row from phenos. if fall conspic according to USDA return TRUE if not 
    return False, if no data NaN"""
    conspic = row["Morphology/Physiology | Fall Conspicuous"]
    if conspic == "Yes":
        return True
    elif conspic == "No":
        return False
    else:
        return np.nan
def mobot_fall_conspic(row):
    """Take row from phenos. if fall conspic according to mobot return true
    if not return false, if no data NaN"""
    conspic = row["Leaf"]
    if conspic == "Colorful, Good Fall" or conspic == "Fragrant, Good Fall" or conspic == "Good Fall":
        return True
    elif conspic == "N/a":
        return np.nan
    else:
        return False
    
def fall_conspic(row):
    """Combine data from usda and mobot to determine if tree is conspicuous in fall"""
    mobot = mobot_fall_conspic(row)
    usda = usda_fall_conspic(row)
    if usda != usda and mobot != mobot:
        return np.nan
    elif usda is True or mobot is True:
        return True
    else:
        return False
    
```

```python
phenos["Fall_Conspic"] = phenos.apply(lambda row: fall_conspic(row), axis=1)
```
