```python
from sklearn.cluster import dbscan
from sklearn.cluster import HDBSCAN
import geopandas as gpd
import folium
import tempfile
import webbrowser
import os
```

```python
trees = gpd.read_file("/Users/prince/philly-tree-mapper/data/processed/philly_trees_located.geojson").to_crs(epsg=3857)
hoods = gpd.read_file("/Users/prince/philly-tree-mapper/data/raw/MUSA_zillow_neighborhoods.geojson").to_crs(epsg=3857)
```

```python
trees["scientific_name"] = trees["scientific_name"].str.lower()
# create X Y columns in tree gdf
trees["x"] = trees.geometry.x
trees["y"] = trees.geometry.y
```

```python
def get_species(scientific_name: str, gdf=trees):
    species_gdf = gdf.loc[gdf["scientific_name"] == scientific_name]
    return species_gdf
```

```python
def do_dbscan(gdf, eps=100, min_samples=20):
    cores, labels = dbscan(gdf[["x", "y"]], eps=eps, min_samples=min_samples)
    gdf["label"] = labels
    return gdf
```

```python
def show_in_browser(m):
    """Open a folium map in the default browser without saving to project dir."""
    tmp = tempfile.NamedTemporaryFile(suffix=".html", delete=False)
    tmp.close()
    m.save(tmp.name)
    webbrowser.open(f"file://{tmp.name}")
```

```python
do_dbscan(get_species("ilex opaca"))
```

```python
species = get_species("acer rubrum")
clustered = do_dbscan(species, eps=150, min_samples=10)
```

```python
clustered
```

```python
m = hoods.explore(
    style_kwds={"weight": 1, "color": "black", "fillColor": "none"},
    tiles="cartodbpositron"
)
clustered[clustered["label"] == -1].explore(
    m=m, color="gray", marker_kwds={"radius": 2},
    style_kwds={"stroke": False, "fillOpacity": 0.7},
)
clustered[clustered["label"] != -1].explore(
    m=m, column="label", categorical=True, cmap="tab20",
    marker_kwds={"radius": 3}, style_kwds={"stroke": False, "fillOpacity": 0.8},
)
```

```python
show_in_browser(m)
```
