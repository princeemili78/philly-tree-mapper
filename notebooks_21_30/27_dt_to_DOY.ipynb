```python
import pandas as pd
import numpy as np
import re
pd.set_option('display.max_columns', None)
```

```python
def parse_dt_index(cell):
    if pd.isna(cell):
        return np.nan
    
    dates = re.findall(r"'(\d[^']*)'", cell)
    return pd.DatetimeIndex(pd.to_datetime(dates))
```

```python
phenos = pd.read_csv("/Users/prince/philly-tree-mapper/data/processed/philly_species_pheno_deaggregated.csv")
phenos['bloom_range'] = phenos['bloom_range'].apply(parse_dt_index)
phenos['fruit_range'] = phenos['fruit_range'].apply(parse_dt_index)
```

```python
print(type(phenos.loc[2, "bloom_range"]))
```

```python
# Function to turn DatetimeIndex to DOY list
def dt_to_doy(dti):
    if isinstance(dti, pd.DatetimeIndex):
        return dti.day_of_year.tolist()
    else:
        return np.nan
```

```python
# turn phenos dt indexes to DOY lists
phenos["bloom_range_DOY"] = phenos["bloom_range"].apply(lambda dti: dt_to_doy(dti))
phenos["fruit_range_DOY"] = phenos["fruit_range"].apply(lambda dti: dt_to_doy(dti))
```
